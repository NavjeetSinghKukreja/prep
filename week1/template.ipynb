```python
from transformers import AutoTokenizer
tok = AutoTokenizer.from_pretrained("Qwen/Qwen3-0.6B", token=False)
messages = [
    {"role": "system", "content": "You answer in one word."},
    {"role": "user", "content": "Capital of France?"},
]
text = tok.apply_chat_template(messages, tokenize=False)  # the string the model reads
ids  = tok.apply_chat_template(messages, tokenize=True)   # the integers it actually gets
```

```python

messages = [{"role": "system", "content": "You answer in one word."}]

messages.append({"role": "user", "content": "Capital of France?"}) 
messages.append({"role": "assistant", "content": "Paris"}) 

messages.append({"role": "user", "content": "And Germany?"})
messages.append({"role": "assistant", "content": "Berlin"}) 

text = tok.apply_chat_template(messages, tokenize=False)
print(text)
print(repr(text))
```
