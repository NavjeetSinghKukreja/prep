```python
import os
from dotenv import load_dotenv
from openai import AsyncOpenAI
load_dotenv()
client = AsyncOpenAI(base_url="https://openrouter.ai/api/v1",
                api_key=os.environ["OPENROUTER_API_KEY"])

MODEL = "qwen/qwen3.8-flash"


async def chat(messages):
    resp = await client.chat.completions.create(
    model= MODEL,
    messages = messages,
    temperature=1,
    )
    return resp.choices[0].message.content

test = [
    {"role": "system", "content": "You answer in one word."},
    {"role": "user", "content": "Capital of France?"},
]

print(await chat(test))
```

```python
import os
from dotenv import load_dotenv
from openai import AsyncOpenAI
load_dotenv()
client = AsyncOpenAI(base_url="https://openrouter.ai/api/v1",
                api_key=os.environ["OPENROUTER_API_KEY"])

MODEL = "qwen/qwen3.8-flash"
PROMPT = "Name one random animal. Reply with just the word."

async def real_call(prompt, temperature):
    resp = await client.chat.completions.create(
    model= MODEL,
    messages=[{"role": "user", "content": prompt}],
    temperature= temperature,
    )
    return resp.choices[0].message.content

```

```python
sem = asyncio.Semaphore(5)

async def limited_call(i, temperature):
    async with sem:
        for attempt in range(3):
            try:
                r = {"id": i, "response": await real_call(PROMPT), "error": None, "attempts": attempt+1, "model": MODEL, "prompt": PROMPT, "temperature": temperature}
                break
            except Exception as e:
                if attempt != 2:
                    await asyncio.sleep(2 ** attempt)
                r = {"id": i, "response": None, "error": f"error: {e}", "attempts": attempt+1, "model": MODEL, "prompt": PROMPT, "temperature": temperature}
        return r

start = time.time()
results = await asyncio.gather(*(limited_call(i) for i in range(1)))
print (results)
print(f"{time.time() - start:.1f}s")
```
